```python
# Lesson 8: Filling Missing Numerical Values
# Step 3: Fill missing values using training medians

import pandas as pd

# Identify numerical columns that still contain missing values
columns_to_fill = df.select_dtypes(
    include=["number"]
).columns[
    df.select_dtypes(include=["number"]).isna().any()
].tolist()

# Use records before 2012 to calculate training medians
training_rows = df["saleYear"] < 2012

# Store median values for future use
numerical_medians = {}

for column in columns_to_fill:

    # Calculate the median from training records only
    median_value = df.loc[training_rows, column].median()

    # Use zero if the training column has no available values
    if pd.isna(median_value):
        median_value = 0

    # Save the median
    numerical_medians[column] = median_value

    # Fill missing values
    df[column] = df[column].fillna(median_value)

# Update features and target
X = df.drop("SalePrice", axis=1)
y = df["SalePrice"]

# Display results
print("NUMERICAL IMPUTATION RESULTS")
print("----------------------------")

print("Columns filled:", len(columns_to_fill))

print("\nMEDIAN VALUES USED")
print("------------------")

for column, median in numerical_medians.items():
    print(column, ":", median)

print("\nREMAINING MISSING NUMERICAL VALUES")
print("----------------------------------")

print(df.select_dtypes(include=["number"]).isna().sum().sum())

print("\nFINAL DATASET")
print("-------------")

print("Total records:", len(df))
print("Total features:", X.shape[1])

display(
    df[
        [
            "MachineHoursCurrentMeter",
            "MachineHoursCurrentMeter_is_missing"
        ]
    ].head(10)
)
```

```python
# Lesson 10: Fitting a Machine Learning Model
# Step 1: Prepare training and validation data

import pandas as pd

# Separate training and validation records
train_df = df[df["saleYear"] < 2012].copy()
valid_df = df[df["saleYear"] == 2012].copy()

# Separate features and target
X_train = train_df.drop("SalePrice", axis=1)
y_train = train_df["SalePrice"]

X_valid = valid_df.drop("SalePrice", axis=1)
y_valid = valid_df["SalePrice"]

# Display dataset information
print("TRAINING AND VALIDATION REPORT")
print("------------------------------")

print("Training records:", len(X_train))
print("Validation records:", len(X_valid))

print("\nFEATURE INFORMATION")
print("-------------------")

print("Training features:", X_train.shape[1])
print("Validation features:", X_valid.shape[1])

print("\nTARGET INFORMATION")
print("------------------")

print("Training target:", y_train.name)
print("Validation target:", y_valid.name)

# Verify our split
assert len(X_train) > 0, "Training dataset is empty!"
assert len(X_valid) > 0, "Validation dataset is empty!"

assert X_train.columns.equals(X_valid.columns), (
    "Training and validation features do not match!"
)

print("\nSUCCESS: Training and validation datasets created!")

# Display sample training data
display(X_train.head())
```

```python
# Lesson 10: Fitting a Machine Learning Model
# Step 2: Train RandomForestRegressor

from sklearn.ensemble import RandomForestRegressor
import time

# Create our machine learning model
model = RandomForestRegressor(
    n_estimators=20,
    max_samples=10000,
    random_state=42,
    n_jobs=-1
)

print("TRAINING OUR BULLDOZER MODEL")
print("----------------------------")

# Record training start time
start_time = time.time()

# Train the model
model.fit(X_train, y_train)

# Calculate training time
training_time = time.time() - start_time

print("Model training completed successfully!")

print("\nMODEL INFORMATION")
print("-----------------")

print("Algorithm:", type(model).__name__)
print("Decision trees:", model.n_estimators)
print("Training records available:", len(X_train))
print("Features used:", X_train.shape[1])
print("Training time:", round(training_time, 2), "seconds")

print("\nMODEL STATUS")
print("------------")
print("Model fitted successfully!")
```

```python
# Lesson 10: Fitting a Machine Learning Model
# Step 3: Make bulldozer price predictions

import pandas as pd
from sklearn.metrics import mean_absolute_error

# Make predictions using validation data
valid_predictions = model.predict(X_valid)

# Compare actual prices with predictions
results = pd.DataFrame({
    "Actual Price": y_valid.to_numpy(),
    "Predicted Price": valid_predictions
})

# Calculate prediction differences
results["Absolute Error"] = (
    results["Actual Price"] - results["Predicted Price"]
).abs()

print("BULLDOZER PRICE PREDICTIONS")
print("---------------------------")

# Display the first 10 predictions
display(results.head(10).round(2))

# Calculate average prediction error
mae = mean_absolute_error(
    y_valid,
    valid_predictions
)

print("\nMODEL PERFORMANCE")
print("-----------------")

print("Validation records:", len(X_valid))
print("Predictions generated:", len(valid_predictions))
print("Mean Absolute Error: ${:,.2f}".format(mae))

# Display a single prediction
print("\nEXAMPLE PREDICTION")
print("------------------")

print(
    "Actual Price: ${:,.2f}".format(
        results.iloc[0]["Actual Price"]
    )
)

print(
    "Predicted Price: ${:,.2f}".format(
        results.iloc[0]["Predicted Price"]
    )
)

print("\nSUCCESS: Model predictions completed!")
```

```python
# Lesson 13: Custom Evaluation Function
# Step 1: Create a reusable evaluation function

import numpy as np

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_log_error,
    r2_score
)

# Create our custom evaluation function
def evaluate_model(y_true, y_pred):

    # Calculate Mean Absolute Error
    mae = mean_absolute_error(y_true, y_pred)

    # Calculate R-squared
    r2 = r2_score(y_true, y_pred)

    # Check that RMSLE can be calculated
    if np.any(np.asarray(y_true) < 0) or np.any(np.asarray(y_pred) < 0):
        raise ValueError("RMSLE requires nonnegative values.")

    # Calculate Root Mean Squared Logarithmic Error
    rmsle = np.sqrt(
        mean_squared_log_error(y_true, y_pred)
    )

    # Store evaluation results in a dictionary
    results = {
        "MAE": mae,
        "R2": r2,
        "RMSLE": rmsle
    }

    return results


print("CUSTOM EVALUATION FUNCTION")
print("--------------------------")
print("Function created successfully!")

# Test our function with simple example data
actual_prices = [10000, 20000, 30000]
predicted_prices = [11000, 19000, 32000]

test_results = evaluate_model(
    actual_prices,
    predicted_prices
)

print("\nEXAMPLE EVALUATION RESULTS")
print("--------------------------")

for metric, value in test_results.items():
    print(metric + ":", round(value, 4))
```

```python
# Lesson 13: Custom Evaluation Function
# Step 2: Evaluate our actual bulldozer model

# Generate validation predictions
valid_predictions = model.predict(X_valid)

# Evaluate our model using the custom function
validation_scores = evaluate_model(
    y_valid,
    valid_predictions
)

print("BULLDOZER MODEL EVALUATION")
print("--------------------------")

print("Validation records:", len(y_valid))

print("\nMODEL PERFORMANCE")
print("-----------------")

print(
    "Mean Absolute Error: ${:,.2f}".format(
        validation_scores["MAE"]
    )
)

print(
    "R² Score:",
    round(validation_scores["R2"], 4)
)

print(
    "RMSLE:",
    round(validation_scores["RMSLE"], 4)
)

print("\nEVALUATION COMPLETE!")
```

```python
# Lesson 16: Improving Hyperparameters
# Step 1: Create an improved hyperparameter search

import numpy as np

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import make_scorer, mean_squared_log_error

# Create a custom RMSLE scoring function
def calculate_rmsle(y_true, y_pred):
    y_pred = np.maximum(y_pred, 0)

    return np.sqrt(
        mean_squared_log_error(y_true, y_pred)
    )

# Tell Scikit-learn that LOWER RMSLE is better
rmsle_scorer = make_scorer(
    calculate_rmsle,
    greater_is_better=False
)

# Define our improved hyperparameter options
improved_params = {
    "n_estimators": [20, 40, 60],
    "max_samples": [10000, 20000, 30000],
    "max_depth": [None, 20],
    "min_samples_split": [2, 4],
    "min_samples_leaf": [1, 2]
}

# Create our Random Forest
improved_rf = RandomForestRegressor(
    random_state=42,
    n_jobs=-1
)

# Create our improved randomized search
improved_search = RandomizedSearchCV(
    estimator=improved_rf,
    param_distributions=improved_params,
    n_iter=6,
    cv=time_cv,
    scoring=rmsle_scorer,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

print("IMPROVED HYPERPARAMETER SEARCH")
print("------------------------------")
print("Search created successfully!")
print("Combinations to test: 6")
print("Cross-validation folds: 3")
print("Expected CV model fits: 18")
print("Scoring metric: RMSLE")
print("Training records:", len(X_train_ordered))
```

```python
# Lesson 17: Preprocessing Our Data
# Step 2: Build reusable preprocessing rules

import pandas as pd
import numpy as np

class BulldozerPreprocessor:

    def __init__(self):
        self.category_mappings = {}
        self.numerical_medians = {}
        self.feature_columns = None

    def prepare_dates(self, data):
        data = data.copy()

        dates = pd.to_datetime(data["saledate"])

        data["saleYear"] = dates.dt.year
        data["saleMonth"] = dates.dt.month
        data["saleDay"] = dates.dt.day
        data["saleDayOfWeek"] = dates.dt.dayofweek
        data["saleDayOfYear"] = dates.dt.dayofyear
        data["saleQuarter"] = dates.dt.quarter

        data["equipmentAge"] = (
            data["saleYear"] - data["YearMade"]
        )

        data = data.drop("saledate", axis=1)

        return data

    def fit(self, data):
        data = self.prepare_dates(data)

        # Learn category mappings from training data only
        for column in data.select_dtypes(
            include=["object", "category", "string"]
        ).columns:

            categories = (
                data[column]
                .dropna()
                .astype(str)
                .unique()
            )

            self.category_mappings[column] = {
                category: index + 1
                for index, category in enumerate(categories)
            }

        # Learn numerical medians from training data only
        numerical_columns = data.select_dtypes(
            include=["number"]
        ).columns

        for column in numerical_columns:
            median = data[column].median()

            self.numerical_medians[column] = (
                median if pd.notna(median) else 0
            )

        self.feature_columns = data.columns.tolist()

        return self

    def transform(self, data):
        data = self.prepare_dates(data)

        data = data.reindex(
            columns=self.feature_columns
        )

        for column, mapping in self.category_mappings.items():
            data[column] = (
                data[column]
                .astype("string")
                .map(mapping)
                .fillna(0)
                .astype(int)
            )

        for column, median in self.numerical_medians.items():
            data[column] = data[column].fillna(median)

        return data

    def fit_transform(self, data):
        return self.fit(data).transform(data)


print("BULLDOZER PREPROCESSOR")
print("----------------------")
print("Reusable preprocessing class created!")
print("Training-only category mappings: Ready")
print("Training-only numerical medians: Ready")
print("Date feature extraction: Ready")

print("\nSTEP 2 COMPLETE!")
```
